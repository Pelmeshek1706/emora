# head_movement


```python
import openwillis.face as owf

print("has head_movement:", hasattr(owf, "head_movement"))
owf.head_movement
```

```python
import openwillis.face as owf
# 11.7mb video 1080p MPEG-4 AAC, H.264, stereo, duration 21s
# frames_per_second=1 - 14.4s
# frames_per_second=3 - 1m4s - for 30fps video result will appears only 3 times - every 10 frames
# frames_per_second=10 - 1m43s
# frames_per_second=15 - 2m34s
# frames_per_second=25 - 5m17s
# frames_per_second=30 - 5m7s (3m33s with bounding boxes + 1m13s for bbox)

head_movement_df, summary_df = owf.head_movement(
	video_path='video.mov',
	frames_per_second=30, # for 30fps video result will appears only 3 times - every 10 frames
	normalize_by_bb_size=False,
	bbox_list=[],
	padding_percent=0.1,
)

print("head_movement_df shape:", head_movement_df.shape)
print("summary_df shape:", summary_df.shape)
# display(head_movement_df.head())
# display(summary_df)
```

```python
# 11.7mb video 1080p MPEG-4 AAC, H.264, stereo, duration 21s
# bbox computed 1m13s
# frames_per_second=1 - 10.7s
# frames_per_second=3 - 24.5s
# frames_per_second=10 - 1m12s
# frames_per_second=15 - 1m49s
# frames_per_second=25 - 3m51s
# frames_per_second=30 - 3m33s

head_movement_df, summary_df = owf.head_movement(
    video_path="video.mov",
    frames_per_second=30,
    normalize_by_bb_size=False,
    bbox_list=bbox_list,
    padding_percent=0.1,
)
```
